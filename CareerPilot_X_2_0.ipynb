```python
# ================================================================
# 🚀 CAREERPILOT-X 2.0
# AI CAREER OPERATING SYSTEM
# COMPLETE MAIN CODE
# ================================================================

!pip install -q requests pandas

import requests
import pandas as pd
import re
import random
from datetime import datetime


# ================================================================
# 🔐 SERPAPI
# ================================================================

print("=" * 75)
print("🚀 CAREERPILOT-X 2.0")
print("AI CAREER OPERATING SYSTEM")
print("=" * 75)

print("\n🔐 SerpApi is required only for LIVE JOB SEARCH.")
print("⚠️ Enter your API key only in this input box.\n")

SERPAPI_KEY = input(
    "🔑 Enter your SerpApi API Key: "
).strip()


# ================================================================
# 👤 STUDENT PROFILE
# ================================================================

print("\n" + "=" * 75)
print("👤 STUDENT PROFILE")
print("=" * 75)

student_name = input(
    "\n👤 Enter your name: "
).strip()

student = {

    "name": student_name,

    "year": "3rd BTech",

    "degree": "BTech CSE-AIML",

    "location": "India",

    "skills": [
        "Python",
        "Java",
        "Machine Learning",
        "NLP",
        "Generative AI",
        "RAG",
        "Git"
    ],

    "projects": [
        "Fake News Detection",
        "CareerPath AI",
        "EcoVision",
        "CareerPilot-X",
        "ScamShield AI"
    ],

    "experience": [
        "GenAI Internship",
        "Hackathon Participation"
    ]
}


# ================================================================
# 🎯 CAREER DATABASE
# ================================================================

CAREERS = {

    "AI/ML Engineer": [
        "Python",
        "Machine Learning",
        "Statistics",
        "NumPy",
        "Pandas",
        "Deep Learning",
        "NLP",
        "TensorFlow",
        "PyTorch",
        "Generative AI",
        "RAG"
    ],

    "Data Scientist": [
        "Python",
        "Statistics",
        "Pandas",
        "NumPy",
        "SQL",
        "Machine Learning",
        "Data Visualization"
    ],

    "Data Analyst": [
        "Python",
        "SQL",
        "Excel",
        "Statistics",
        "Pandas",
        "Power BI",
        "Data Visualization"
    ],

    "Software Developer": [
        "Python",
        "Java",
        "Data Structures",
        "Algorithms",
        "SQL",
        "Git",
        "APIs"
    ],

    "Web Developer": [
        "HTML",
        "CSS",
        "JavaScript",
        "React",
        "APIs",
        "SQL",
        "Git"
    ],

    "Cybersecurity Analyst": [
        "Networking",
        "Linux",
        "Python",
        "Cybersecurity",
        "OWASP",
        "SIEM",
        "Ethical Hacking"
    ],

    "Cloud Engineer": [
        "AWS",
        "Azure",
        "Linux",
        "Networking",
        "Docker",
        "Kubernetes",
        "Python"
    ],

    "DevOps Engineer": [
        "Linux",
        "Git",
        "Docker",
        "Kubernetes",
        "CI/CD",
        "AWS",
        "Python"
    ],

    "Business Analyst": [
        "Excel",
        "SQL",
        "Business Analysis",
        "Communication",
        "Power BI",
        "Requirements Analysis"
    ],

    "UI/UX Designer": [
        "Figma",
        "UI Design",
        "UX Design",
        "Wireframing",
        "Prototyping",
        "User Research"
    ]
}


# ================================================================
# 🗺️ LEARNING ROADMAP
# ================================================================

ROADMAP = {

    "AI/ML Engineer": [
        "Python",
        "NumPy",
        "Pandas",
        "Statistics",
        "Machine Learning",
        "Deep Learning",
        "NLP",
        "Generative AI",
        "RAG",
        "MLOps",
        "Deployment"
    ],

    "Data Scientist": [
        "Python",
        "Statistics",
        "SQL",
        "Pandas",
        "Machine Learning",
        "Data Visualization",
        "Feature Engineering",
        "Projects"
    ],

    "Data Analyst": [
        "Excel",
        "SQL",
        "Statistics",
        "Python",
        "Pandas",
        "Power BI",
        "Data Visualization",
        "Projects"
    ],

    "Software Developer": [
        "Python / Java",
        "Data Structures",
        "Algorithms",
        "SQL",
        "APIs",
        "Git",
        "Projects",
        "Interview Preparation"
    ],

    "Web Developer": [
        "HTML",
        "CSS",
        "JavaScript",
        "React",
        "Backend",
        "Databases",
        "APIs",
        "Deployment"
    ],

    "Cybersecurity Analyst": [
        "Networking",
        "Linux",
        "Cybersecurity",
        "OWASP",
        "Ethical Hacking",
        "SIEM",
        "Incident Response"
    ],

    "Cloud Engineer": [
        "Linux",
        "Networking",
        "AWS / Azure",
        "Docker",
        "Kubernetes",
        "Cloud Security",
        "Projects"
    ],

    "DevOps Engineer": [
        "Linux",
        "Git",
        "Docker",
        "CI/CD",
        "Kubernetes",
        "Cloud",
        "Automation"
    ],

    "Business Analyst": [
        "Business Fundamentals",
        "Excel",
        "SQL",
        "Business Analysis",
        "Power BI",
        "Requirements",
        "Case Studies"
    ],

    "UI/UX Designer": [
        "Design Fundamentals",
        "Figma",
        "UI Design",
        "UX Research",
        "Wireframing",
        "Prototyping",
        "Portfolio"
    ]
}


# ================================================================
# 🎤 INTERVIEW QUESTION DATABASE
# ================================================================

INTERVIEW_QUESTIONS = {

    "AI/ML Engineer": [

        "What is supervised learning?",

        "What is the difference between classification and regression?",

        "What is overfitting and how can you prevent it?",

        "Explain precision, recall and F1-score.",

        "What is feature engineering?",

        "What is deep learning?",

        "What is NLP?",

        "What is Generative AI?",

        "What is RAG?",

        "How would you deploy a machine learning model?"
    ],

    "Data Scientist": [

        "What is hypothesis testing?",

        "How do you handle missing values?",

        "Explain bias and variance.",

        "What is feature engineering?",

        "How do you evaluate a machine learning model?"
    ],

    "Data Analyst": [

        "What is a SQL JOIN?",

        "How do you find duplicate records?",

        "What is a pivot table?",

        "What is the difference between mean and median?",

        "How do you communicate data insights?"
    ],

    "Software Developer": [

        "Explain object-oriented programming.",

        "What are data structures?",

        "What is time complexity?",

        "What is an API?",

        "What is Git?"
    ],

    "Web Developer": [

        "What is HTML?",

        "What is CSS?",

        "What is JavaScript?",

        "What is the DOM?",

        "What is responsive design?"
    ],

    "Cybersecurity Analyst": [

        "What is phishing?",

        "What is the CIA triad?",

        "What is SQL injection?",

        "What is a firewall?",

        "What is incident response?"
    ],

    "Cloud Engineer": [

        "What is cloud computing?",

        "What is AWS?",

        "What is IAM?",

        "What is Docker?",

        "What is Kubernetes?"
    ],

    "DevOps Engineer": [

        "What is CI/CD?",

        "What is Docker?",

        "What is Kubernetes?",

        "What is infrastructure as code?",

        "Why is monitoring important?"
    ],

    "Business Analyst": [

        "What is requirements gathering?",

        "What is a KPI?",

        "How do you handle conflicting requirements?",

        "How can SQL help a business analyst?",

        "What is business process analysis?"
    ],

    "UI/UX Designer": [

        "What is the difference between UI and UX?",

        "What is a user persona?",

        "What is wireframing?",

        "What is prototyping?",

        "What is user research?"
    ]
}


# ================================================================
# 🧠 UTILITY
# ================================================================

def clean(text):

    return re.sub(
        r"[^a-z0-9+#. ]",
        " ",
        str(text).lower()
    )


def skill_similarity(
    user_skills,
    career_skills
):

    user = set(
        clean(skill)
        for skill in user_skills
    )

    matched = []
    missing = []

    for skill in career_skills:

        if clean(skill) in user:

            matched.append(skill)

        else:

            missing.append(skill)

    score = round(
        len(matched) /
        max(len(career_skills), 1)
        * 100,
        2
    )

    return score, matched, missing


# ================================================================
# 🎯 FEATURE 1 — CAREER MATCH ENGINE
# ================================================================

def career_match():

    results = []

    for career, skills in CAREERS.items():

        score, matched, missing = skill_similarity(
            student["skills"],
            skills
        )

        results.append({

            "Career": career,

            "Match Score": score,

            "Matched Skills": len(matched),

            "Missing Skills": len(missing)

        })

    df = pd.DataFrame(results)

    return df.sort_values(
        "Match Score",
        ascending=False
    ).reset_index(drop=True)


# ================================================================
# 🧠 FEATURE 2 — SKILL GAP
# ================================================================

def skill_gap(career):

    return skill_similarity(
        student["skills"],
        CAREERS[career]
    )


# ================================================================
# 📊 FEATURE 4 — READINESS SCORE
# ================================================================

def readiness_score(career):

    skill_score, _, _ = skill_gap(career)

    project_score = min(
        len(student["projects"]) * 5,
        25
    )

    experience_score = min(
        len(student["experience"]) * 7,
        15
    )

    profile_score = 10

    score = round(
        skill_score * 0.50
        + project_score
        + experience_score
        + profile_score,
        2
    )

    return min(score, 100)


# ================================================================
# 🤖 FEATURE 5 — DIGITAL TWIN
# ================================================================

def digital_twin(career):

    score, matched, missing = skill_gap(career)

    project_score = min(
        len(student["projects"]) * 5,
        25
    )

    experience_score = min(
        len(student["experience"]) * 7,
        15
    )

    final_score = round(
        score * 0.60
        + project_score
        + experience_score,
        2
    )

    final_score = min(
        final_score,
        100
    )

    if final_score >= 80:

        level = "Job Ready"

    elif final_score >= 60:

        level = "Nearly Ready"

    elif final_score >= 40:

        level = "Developing"

    else:

        level = "Beginner"

    return {

        "score": final_score,

        "level": level,

        "matched": matched,

        "missing": missing

    }


# ================================================================
# 🧪 FEATURE 6 — WHAT-IF LAB
# ================================================================

def what_if_analysis(
    career,
    extra_skills
):

    current_score, _, _ = skill_gap(
        career
    )

    future_skills = (
        student["skills"]
        + extra_skills
    )

    future_score, matched, missing = skill_similarity(
        future_skills,
        CAREERS[career]
    )

    return {

        "current": current_score,

        "future": future_score,

        "improvement":
            round(
                future_score -
                current_score,
                2
            ),

        "matched": matched,

        "missing": missing

    }


# ================================================================
# 🌐 FEATURE 7 — LIVE JOB SEARCH
# ================================================================

def search_jobs(query):

    if not SERPAPI_KEY:

        return []

    params = {

        "engine": "google_jobs",

        "q": query,

        "location": "India",

        "hl": "en",

        "api_key": SERPAPI_KEY

    }

    try:

        response = requests.get(

            "https://serpapi.com/search.json",

            params=params,

            timeout=60

        )

        response.raise_for_status()

        data = response.json()

        jobs = []

        for job in data.get(
            "jobs_results",
            []
        ):

            title = job.get(
                "title",
                ""
            ).strip()

            company = job.get(
                "company_name",
                ""
            ).strip()

            description = job.get(
                "description",
                ""
            ).strip()

            apply_options = job.get(
                "apply_options",
                []
            )

            apply_url = ""

            if apply_options:

                apply_url = apply_options[0].get(
                    "link",
                    ""
                )

            if not title or not company:

                continue

            jobs.append({

                "Title": title,

                "Company": company,

                "Location":
                    job.get(
                        "location",
                        "Not specified"
                    ),

                "Type":
                    job.get(
                        "detected_extensions",
                        {}
                    ).get(
                        "schedule_type",
                        "Not specified"
                    ),

                "Posted":
                    job.get(
                        "detected_extensions",
                        {}
                    ).get(
                        "posted_at",
                        "Not specified"
                    ),

                "Description":
                    description,

                "Apply URL":
                    apply_url

            })

        return jobs

    except Exception as e:

        print(
            "\n⚠️ Live job search could not be completed."
        )

        print(
            "Reason:",
            str(e)
        )

        return []


# ================================================================
# 🧬 FEATURE 8 — OPPORTUNITY DNA
# ================================================================

def opportunity_dna(
    job,
    career
):

    required = CAREERS[career]

    text = clean(
        job["Title"]
        + " "
        + job["Description"]
    )

    matched = []
    missing = []

    for skill in required:

        if clean(skill) in text:

            matched.append(skill)

        else:

            missing.append(skill)

    fit = round(

        len(matched) /
        max(len(required), 1)
        * 100,

        2
    )

    return {

        "fit": fit,

        "matched": matched,

        "missing": missing

    }


# ================================================================
# 📄 FEATURE 9 — RESUME ANALYZER
# ================================================================

def resume_analyzer(
    resume_text,
    career
):

    required = CAREERS[career]

    text = clean(
        resume_text
    )

    found = []
    missing = []

    for skill in required:

        if clean(skill) in text:

            found.append(skill)

        else:

            missing.append(skill)

    sections = {

        "Education":
            bool(
                re.search(
                    r"education|degree|btech|bachelor",
                    text
                )
            ),

        "Skills":
            bool(
                re.search(
                    r"skills|technical skills",
                    text
                )
            ),

        "Projects":
            bool(
                re.search(
                    r"projects|project",
                    text
                )
            ),

        "Experience":
            bool(
                re.search(
                    r"experience|internship",
                    text
                )
            ),

        "Certifications":
            bool(
                re.search(
                    r"certification|certificate",
                    text
                )
            )

    }

    keyword_score = round(

        len(found) /
        max(len(required), 1)
        * 100,

        2
    )

    section_score = round(

        sum(sections.values()) /
        len(sections)
        * 100,

        2
    )

    overall = round(

        keyword_score * 0.70
        + section_score * 0.30,

        2
    )

    return {

        "overall": overall,

        "keyword": keyword_score,

        "sections": section_score,

        "found": found,

        "missing": missing

    }


# ================================================================
# 🎤 FEATURE 10 — INTERVIEW SIMULATOR
# ================================================================

def run_careerpilot_interview(
    career
):

    questions = INTERVIEW_QUESTIONS.get(
        career,
        []
    )

    if not questions:

        print(
            "❌ No questions available for:",
            career
        )

        return

    print("\n")
    print("=" * 75)
    print(
        "🎤 CAREERPILOT-X 2.0 — "
        "INTERVIEW SIMULATOR"
    )
    print("=" * 75)

    print(
        f"\n🎯 Target Career: {career}"
    )

    selected_questions = random.sample(

        questions,

        min(
            5,
            len(questions)
        )

    )

    total_score = 0

    for number, question in enumerate(

        selected_questions,

        start=1

    ):

        print("\n" + "-" * 75)

        print(
            f"❓ QUESTION "
            f"{number}/"
            f"{len(selected_questions)}"
        )

        print(
            f"\n{question}"
        )

        answer = input(
            "\n✍️ Your Answer: "
        ).strip()

        word_count = len(
            answer.split()
        )

        if word_count >= 50:

            score = 10

            feedback = (
                "Excellent explanation. "
                "Add a practical example."
            )

        elif word_count >= 30:

            score = 8

            feedback = (
                "Good explanation. "
                "Try adding a real-world example."
            )

        elif word_count >= 15:

            score = 6

            feedback = (
                "Good start. "
                "Add more technical details."
            )

        elif word_count >= 7:

            score = 4

            feedback = (
                "Basic answer. "
                "Explain the concept more clearly."
            )

        else:

            score = 2

            feedback = (
                "Answer is too short. "
                "Explain the concept step-by-step."
            )

        total_score += score

        print(
            f"\n📊 Question Score: "
            f"{score}/10"
        )

        print(
            f"💡 AI Feedback: "
            f"{feedback}"
        )

    final_score = round(

        (
            total_score /
            (
                len(selected_questions)
                * 10
            )
        )
        * 100,

        2
    )

    print("\n")
    print("=" * 75)
    print(
        "🏆 INTERVIEW PERFORMANCE"
    )
    print("=" * 75)

    print(
        f"\n🎯 Final Interview Score: "
        f"{final_score}/100"
    )

    if final_score >= 85:

        print(
            "🟢 EXCELLENT — "
            "Interview Ready"
        )

    elif final_score >= 70:

        print(
            "🟡 STRONG — "
            "Keep Practicing"
        )

    elif final_score >= 50:

        print(
            "🟠 DEVELOPING — "
            "Improve Technical Explanations"
        )

    else:

        print(
            "🔴 BEGINNER — "
            "More Interview Practice Needed"
        )

    print(
        "\n🚀 CareerPilot-X Recommendation:"
    )

    if final_score < 70:

        print(
            "Practice explaining your projects using:"
        )

        print(
            "Problem → Technology → "
            "Solution → Result"
        )

    else:

        print(
            "Start practicing "
            "company-specific interview questions."
        )


# ================================================================
# 📋 FEATURE 11 — APPLICATION TRACKER
# ================================================================

applications = []


def add_application(job):

    applications.append({

        "Company":
            job["Company"],

        "Role":
            job["Title"],

        "Status":
            "Applied",

        "Applied On":
            datetime.now().strftime(
                "%Y-%m-%d"
            ),

        "URL":
            job["Apply URL"]

    })


# ================================================================
# 🎮 FEATURE 12 — CAREER MISSION
# ================================================================

missions = [

    "Complete Python fundamentals",

    "Learn NumPy and Pandas",

    "Build one Machine Learning project",

    "Complete one Generative AI project",

    "Build a GitHub portfolio",

    "Analyze 10 job descriptions",

    "Practice 20 interview questions",

    "Apply to 5 verified opportunities"

]


# ================================================================
# 🚀 RUN CAREERPILOT-X
# ================================================================

print("\n")
print("=" * 75)
print(
    "🚀 CAREERPILOT-X 2.0"
)
print(
    "AI CAREER OPERATING SYSTEM"
)
print("=" * 75)

print(
    f"\n👋 Welcome {student['name']}!"
)

print(
    f"🎓 {student['degree']} | "
    f"{student['year']}"
)

print(
    f"🧠 Current Skills: "
    f"{len(student['skills'])}"
)

print(
    f"🚀 Projects: "
    f"{len(student['projects'])}"
)

print(
    f"💼 Experience Items: "
    f"{len(student['experience'])}"
)


# ================================================================
# FEATURE 1
# ================================================================

print("\n")
print("=" * 75)
print(
    "🎯 FEATURE 1 — CAREER MATCH ENGINE"
)
print("=" * 75)

career_results = career_match()

print(
    career_results.to_string(
        index=False
    )
)

best_career = career_results.iloc[0]["Career"]

best_match_score = career_results.iloc[0]["Match Score"]

print(
    f"\n🏆 BEST MATCH: "
    f"{best_career}"
)

print(
    f"🎯 MATCH SCORE: "
    f"{best_match_score}%"
)


# ================================================================
# FEATURE 2
# ================================================================

print("\n")
print("=" * 75)
print(
    "🧠 FEATURE 2 — SKILL GAP ANALYSIS"
)
print("=" * 75)

score, matched, missing = skill_gap(
    best_career
)

print(
    f"\nCareer: {best_career}"
)

print(
    f"Skill Match: {score}%"
)

print(
    "\n✅ Matched Skills:"
)

for skill in matched:

    print(
        f"   ✓ {skill}"
    )

print(
    "\n❌ Missing Skills:"
)

for skill in missing:

    print(
        f"   → {skill}"
    )


# ================================================================
# FEATURE 3
# ================================================================

print("\n")
print("=" * 75)
print(
    "🗺️ FEATURE 3 — "
    "PERSONALIZED LEARNING ROADMAP"
)
print("=" * 75)

for number, step in enumerate(

    ROADMAP[best_career],

    start=1

):

    print(
        f"{number:02d}. {step}"
    )


# ================================================================
# FEATURE 4
# ================================================================

print("\n")
print("=" * 75)
print(
    "📊 FEATURE 4 — "
    "CAREER READINESS SCORE"
)
print("=" * 75)

readiness = readiness_score(
    best_career
)

print(
    f"\n🔥 Career Readiness: "
    f"{readiness}/100"
)

if readiness >= 80:

    print(
        "🟢 STATUS: JOB READY"
    )

elif readiness >= 60:

    print(
        "🟡 STATUS: NEARLY READY"
    )

else:

    print(
        "🔴 STATUS: SKILL DEVELOPMENT NEEDED"
    )


# ================================================================
# FEATURE 5
# ================================================================

print("\n")
print("=" * 75)
print(
    "🤖 FEATURE 5 — "
    "CAREER DIGITAL TWIN"
)
print("=" * 75)

twin = digital_twin(
    best_career
)

print(
    f"\n🎯 Career: "
    f"{best_career}"
)

print(
    f"🧬 Digital Twin Score: "
    f"{twin['score']}/100"
)

print(
    f"📈 Level: "
    f"{twin['level']}"
)

print(
    "\n🧠 Current Skill Strength:"
)

for skill in twin["matched"]:

    print(
        f"   ✓ {skill}"
    )

print(
    "\n⚠️ Digital Twin Skill Gaps:"
)

for skill in twin["missing"]:

    print(
        f"   → {skill}"
    )


# ================================================================
# FEATURE 6
# ================================================================

print("\n")
print("=" * 75)
print(
    "🧪 FEATURE 6 — "
    "WHAT-IF CAREER LAB"
)
print("=" * 75)

extra_skills = [

    "SQL",
    "NumPy",
    "Pandas",
    "PyTorch",
    "TensorFlow",
    "Deep Learning"

]

whatif = what_if_analysis(

    best_career,

    extra_skills

)

print(
    f"\nCurrent Career Match: "
    f"{whatif['current']}%"
)

print(
    f"After Adding Skills: "
    f"{whatif['future']}%"
)

print(
    f"🚀 Improvement: "
    f"+{whatif['improvement']}%"
)

print(
    "\nSkills Added:"
)

for skill in extra_skills:

    print(
        f"   + {skill}"
    )

print(
    "\nRemaining Skill Gaps:"
)

for skill in whatif["missing"]:

    print(
        f"   → {skill}"
    )


# ================================================================
# FEATURE 7
# ================================================================

print("\n")
print("=" * 75)
print(
    "🌐 FEATURE 7 — LIVE JOB SEARCH"
)
print("=" * 75)

job_query = (
    f"{best_career} "
    f"internship India beginner"
)

print(
    "\n🔎 Searching live opportunities for:"
)

print(
    f"   {job_query}"
)

jobs = search_jobs(
    job_query
)

if jobs:

    print(
        f"\n✅ {len(jobs)} "
        f"source-backed opportunities found."
    )

    for i, job in enumerate(
        jobs[:10],
        start=1
    ):

        print(
            "\n" + "-" * 70
        )

        print(
            f"#{i} {job['Title']}"
        )

        print(
            f"🏢 Company: "
            f"{job['Company']}"
        )

        print(
            f"📍 Location: "
            f"{job['Location']}"
        )

        print(
            f"🕒 Type: "
            f"{job['Type']}"
        )

        print(
            f"📅 Posted: "
            f"{job['Posted']}"
        )

        print(
            f"🔗 Apply: "
            f"{job['Apply URL']}"
        )

else:

    print(
        "\n⚠️ No verified opportunities "
        "were returned."
    )

    print(
        "CareerPilot-X will NOT "
        "create fake job listings."
    )


# ================================================================
# FEATURE 8
# ================================================================

print("\n")
print("=" * 75)
print(
    "🧬 FEATURE 8 — "
    "OPPORTUNITY DNA"
)
print("=" * 75)

if jobs:

    selected_job = jobs[0]

    dna = opportunity_dna(

        selected_job,

        best_career

    )

    print(
        f"\n💼 Opportunity: "
        f"{selected_job['Title']}"
    )

    print(
        f"🏢 Company: "
        f"{selected_job['Company']}"
    )

    print(
        f"\n🎯 Skill Fit: "
        f"{dna['fit']}%"
    )

    print(
        "\n✅ Matching Skills:"
    )

    for skill in dna["matched"]:

        print(
            f"   ✓ {skill}"
        )

    print(
        "\n❌ Missing Skills:"
    )

    for skill in dna["missing"]:

        print(
            f"   → {skill}"
        )

else:

    print(
        "⏭️ Opportunity DNA skipped "
        "because no verified job was returned."
    )


# ================================================================
# FEATURE 9
# ================================================================

print("\n")
print("=" * 75)
print(
    "📄 FEATURE 9 — "
    "RESUME ANALYZER"
)
print("=" * 75)

resume_text = """

BTech CSE-AIML student.

Python Java Machine Learning NLP
Generative AI RAG Git.

Projects include Fake News Detection,
CareerPath AI, EcoVision and CareerPilot-X.

Completed GenAI internship.

Participated in hackathons.

Skills include Python, Java,
Machine Learning, NLP,
Generative AI and RAG.

Education:
BTech CSE-AIML

Projects:
Fake News Detection
CareerPath AI
EcoVision
CareerPilot-X

Experience:
GenAI Internship

Certifications:
Technical certifications
"""

resume_result = resume_analyzer(

    resume_text,

    best_career

)

print(
    f"\n📊 Overall Resume Score: "
    f"{resume_result['overall']}/100"
)

print(
    f"🎯 Career Keyword Score: "
    f"{resume_result['keyword']}%"
)

print(
    f"📑 Resume Sections Score: "
    f"{resume_result['sections']}%"
)

print(
    "\n✅ Resume Skills Found:"
)

for skill in resume_result["found"]:

    print(
        f"   ✓ {skill}"
    )

print(
    "\n⚠️ Skills Missing From Resume:"
)

for skill in resume_result["missing"]:

    print(
        f"   → {skill}"
    )


# ================================================================
# FEATURE 10 — INTERVIEW SIMULATOR
# ================================================================

print("\n")
print("=" * 75)
print(
    "🎤 FEATURE 10 — "
    "INTERVIEW SIMULATOR"
)
print("=" * 75)

print(
    f"\n🎯 Recommended Career: "
    f"{best_career}"
)

choice = input(

    "\nStart Interview Simulator? "
    "(yes/no): "

).strip().lower()

if choice == "yes":

    run_careerpilot_interview(
        best_career
    )

else:

    print(
        "\n⏭️ Interview Simulator skipped."
    )


# ================================================================
# FEATURE 11 — CAREER MISSION
# ================================================================

print("\n")
print("=" * 75)
print(
    "🎮 FEATURE 11 — CAREER MISSION"
)
print("=" * 75)

print(
    "\n🚀 Your Career Mission:"
)

for number, mission in enumerate(

    missions,

    start=1

):

    print(
        f"☐ Mission {number}: "
        f"{mission}"
    )


# ================================================================
# FEATURE 12 — APPLICATION TRACKER
# ================================================================

print("\n")
print("=" * 75)
print(
    "📋 FEATURE 12 — "
    "APPLICATION TRACKER"
)
print("=" * 75)

if jobs:

    add_application(
        jobs[0]
    )

if applications:

    tracker = pd.DataFrame(
        applications
    )

    print(
        tracker.to_string(
            index=False
        )
    )

else:

    print(
        "No applications tracked yet."
    )


# ================================================================
# FEATURE 13 — CAREER COMPARISON
# ================================================================

print("\n")
print("=" * 75)
print(
    "⚖️ FEATURE 13 — "
    "CAREER COMPARISON"
)
print("=" * 75)

print(
    career_results.head(5).to_string(
        index=False
    )
)


# ================================================================
# FEATURE 14 — PROFILE INTELLIGENCE
# ================================================================

print("\n")
print("=" * 75)
print(
    "👤 FEATURE 14 — "
    "PROFILE INTELLIGENCE"
)
print("=" * 75)

print(
    f"\nName: "
    f"{student['name']}"
)

print(
    f"Degree: "
    f"{student['degree']}"
)

print(
    f"Year: "
    f"{student['year']}"
)

print(
    f"Location: "
    f"{student['location']}"
)

print(
    f"Skills: "
    f"{len(student['skills'])}"
)

print(
    f"Projects: "
    f"{len(student['projects'])}"
)

print(
    f"Experience: "
    f"{len(student['experience'])}"
)


# ================================================================
# 🧪 FEATURE 15 — FULL SYSTEM TEST
# ================================================================

print("\n")
print("=" * 75)
print(
    "🧪 FEATURE 15 — "
    "FULL SYSTEM TEST"
)
print("=" * 75)

tests = {

    "Career Match Engine":
        len(career_results) == 10,

    "Skill Gap Analysis":
        best_career in CAREERS,

    "Learning Roadmap":
        best_career in ROADMAP,

    "Career Readiness":
        0 <= readiness <= 100,

    "Career Digital Twin":
        0 <= twin["score"] <= 100,

    "What-If Lab":
        whatif["future"] >= 0,

    "Live Job Search":
        isinstance(jobs, list),

    "Opportunity DNA":
        True,

    "Resume Analyzer":
        0 <= resume_result["overall"] <= 100,

    "Interview Simulator":
        best_career in INTERVIEW_QUESTIONS,

    "Career Mission":
        len(missions) > 0,

    "Application Tracker":
        isinstance(applications, list),

    "Career Comparison":
        len(career_results) >= 5,

    "Profile Intelligence":
        student["name"] != ""

}

passed = 0

for feature, result in tests.items():

    if result:

        print(
            f"✅ {feature}"
        )

        passed += 1

    else:

        print(
            f"❌ {feature}"
        )


# ================================================================
# 🏆 FINAL CAREERPILOT-X REPORT
# ================================================================

print("\n")
print("=" * 75)
print(
    "🏆 CAREERPILOT-X 2.0 FINAL REPORT"
)
print("=" * 75)

print(
    f"\n👤 Student: "
    f"{student['name']}"
)

print(
    f"🎯 Recommended Career: "
    f"{best_career}"
)

print(
    f"📊 Career Match: "
    f"{best_match_score}%"
)

print(
    f"🔥 Career Readiness: "
    f"{readiness}/100"
)

print(
    f"🧬 Digital Twin: "
    f"{twin['score']}/100"
)

print(
    f"📄 Resume Score: "
    f"{resume_result['overall']}/100"
)

print(
    f"🌐 Verified Jobs Found: "
    f"{len(jobs)}"
)

print(
    f"\n🧪 System Test: "
    f"{passed}/{len(tests)} features passed"
)

print("\n" + "=" * 75)

if passed == len(tests):

    print(
        "🟢 CAREERPILOT-X 2.0 "
        "SYSTEM STATUS: ALL CORE FEATURES PASSED"
    )

else:

    print(
        "🟡 CAREERPILOT-X 2.0 "
        "SYSTEM STATUS: PARTIALLY PASSED"
    )

print("=" * 75)

print(
    "\n🚀 CAREERPILOT-X 2.0 EXECUTION COMPLETE!"
)
```
